```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datos = pd.read_csv("Automobile.csv")
```

```python
print(datos.head())
```

```python
#Gráfico de dispersión: mpg y weight

plt.scatter(datos["weight"], datos["mpg"])

plt.xlabel("Peso del automóvil")
plt.ylabel("Consumo (mpg)")
plt.title("Relación entre peso y consumo de combustible")

plt.grid()
plt.show()
```

```python
#interpretacion: Al observar el gráfico se puede apreciar una relación negativa entre ambas variables.
#Esto indica que, en general, a medida que aumenta el peso del automóvil, disminuye el valor de mpg.

```

```python
#Histograma de mpg
plt.hist(datos["mpg"], bins=10)

plt.xlabel("Consumo (mpg)")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución del consumo de combustible")

plt.grid()
plt.show()
```

```python
#interpretacion:

El histograma permite observar cómo se distribuyen los valores de mpg.

En el conjunto de datos:

mínimo: 9 mpg
máximo: 46,6 mpg
promedio: aproximadamente 23,51 mpg
mediana: 23 mpg
```

```python
#conversion de datos para que se puedan representar numericamente en el boxsplot
datos["origin_num"] = datos["origin"].map({
    "usa": 1,
    "europe": 2,
    "japan": 3
})
```

```python
#Boxsplot

plt.boxplot([
    datos["cylinders"],
    datos["origin_num"]
])

plt.xticks([1, 2], ["Cylinders", "Origin"])
plt.ylabel("Valor")
plt.title("Boxplot de cylinders y origin")

plt.grid()
plt.show()
```

```python
#Interpretación del boxsplot

El boxplot permite observar la distribución de los valores y posibles valores atípicos.

Para cylinders, los valores presentes son principalmente:

3, 4, 5, 6 y 8
```

```python
# Gráfico de barras: model_year y origin

#Primero podemos crear una tabla de frecuencias utilizando groupby():
tabla = datos.groupby(["model_year", "origin"]).size().unstack(fill_value=0)

print(tabla)
```

```python
# Luego realizamos el grafico:

tabla.plot(kind="bar")

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución de automóviles según año y origen")

plt.grid()
plt.show()
```

```python
# Interpretacion del grafico de barras:

Este gráfico permite comparar, para cada año de modelo, cuántos automóviles corresponden a cada origen.

Por ejemplo, en los primeros años predominan los automóviles de origen estadounidense, mientras que hacia los últimos años se observa una mayor presencia de automóviles provenientes de Japón y Europa.

Esto permite observar cómo cambia la distribución de los automóviles según el año y el origen.
```
